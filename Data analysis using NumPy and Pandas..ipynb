# Pandas Series


- A Pandas Series is a one‑dimensional labeled array that can hold data of any type 
- (integers, floats, strings, objects).
       
<u> Structure  </u>
- It consists of two main parts:

- Values → The actual data.
- Index  → Labels that identify each element.

🟡 Accessing Data

- By index label → marks['Rank2'] → 92
- By position → marks[1] → 92

🟠 Key Features

- Handles missing data gracefully (NaN).
- Supports vectorized operations (fast calculations).
- Can be created from lists, dictionaries, or NumPy arrays.

  ✨ In short: A Pandas Series is like a column in Excel or a 1D NumPy array with labels, making it powerful for data analysis.

```python
# 1. Creating Pandas Series:
# ● Create a Pandas Series named marks with the following data:
# ➢ Marks: 95, 92, 89, 85, 80
# ➢ Custom Index: 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'
```

```python
#Creating Pandas Series:
marks=pd.Series( [95,92,89,85,80],
index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])

print('pandas series')
print()
print(marks)
```

```python
# 2. Indexing and Slicing:
# ● Use the integer index position to access the mark of the 1st rank student.
# ● Use the loc accessor to retrieve the marks of the top 3 ranks by specifying their index labels.
# ● Use the iloc accessor to retrieve the mark of the 3rd-ranked student.
# ● Apply a boolean mask to filter the Series and retrieve the ranks where the marks are greater than 90.
```

```python
# ● Use the integer index position to access the mark of the 1st rank student.
print(marks.iloc[0])
```

```python
# ● Use the loc accessor to retrieve the marks of the top 3 ranks by specifying their index labels.
print(marks.loc[['Rank1','Rank2','Rank3']])
```

```python
# ● Use the iloc accessor to retrieve the mark of the 3rd-ranked student.
print(marks.iloc[2])
```

```python
# ● Apply a boolean mask to filter the Series and retrieve the ranks where the marks are greater than 90.
print(marks[marks> 90])
```

```python
# 3. Manipulating Series:
# ● Modify the mark of the student with the 1st rank to 100.
# ● Remove the entry corresponding to the last rank from the Series.
# ● Compute the CGPA by dividing each mark in the Series by 10.
```

```python
# ● Modify the mark of the student with the 1st rank to 100.
marks['Rank1']=100
marks
```

```python
# ● Remove the entry corresponding to the last rank from the Series.
marks = marks.drop('Rank5')
marks
```

```python
# ● Compute the CGPA by dividing each mark in the Series by 10.
CGPA=marks/10
print('Marks:\n'+str(marks))
print()
print('CGPA:\n'+str(CGPA))
```

```python
# using dataframe CGPA
df = pd.DataFrame({
    'Marks': marks,
    'CGPA': CGPA
})
df
```

# Pandas DataFrame
## # 1. Creating Pandas DataFrame:

```python
# 1. Creating Pandas DataFrame:

# ● Create a Pandas DataFrame named transactions with the following data: ➢
# TransactionID: 101, 102, 103, 104, 105, 106, 107, 108, 109, 110
# ➢ ProductCategory: 'Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',

# 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
# ➢ Region: 'North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South',
# 'North'
# ➢ Amount: 200, 150, 300, 450, 200, 250, 300, 180, 350, 400
```

🔵 Definition  
- A Pandas DataFrame is a two‑dimensional labeled data structure with rows and columns, similar to an Excel spreadsheet or SQL table.

🟢 Structure

- Rows → Represent individual records/observations.
- Columns → Represent variables/attributes.
- Each column is essentially a Pandas Series.

🟡 Key Features

- Handles heterogeneous data (different types in different columns).
- Supports indexing and slicing for rows and columns.

```python
data={
    "TransactionID" : [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "ProductCategory": ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
                        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    "Region"         : ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South','North'],
    "Amount"        :[ 200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
    
}
df_data=pd.DataFrame(data)
df_data
```

## 3. Manipulating the DataFrame:

```python
# 3. Manipulating the DataFrame:
# ● Modify the 'Amount' for the transaction with TransactionID 102 to 165.
# ● Add a new column 'Discount' by calculating 10% of the 'Amount'.
# ● Remove the row with TransactionID 109.
# ● Delete the 'Discount' column from the DataFrame.
```

```python
# ● Modify the 'Amount' for the transaction with TransactionID 102 to 165.
import pandas as pd

# 1. Modify 'Amount' for TransactionID 102 to 165
df_data.loc[df_data['TransactionID'] == 102, 'Amount'] = 165


print(df_data)
```

```python
# 2. Add a new column 'Discount' (10% of Amount)
df_data['Discount'] = df_data['Amount'] * 0.10
print(df_data)
```

```python
# 3. Remove the row with TransactionID 109
df_data = df_data[df_data['TransactionID'] != 109]
print(df_data)
```

```python
# 4. Delete the 'Discount' column
df_data = df_data.drop(columns=['Discount'])
print(df_data)
```
